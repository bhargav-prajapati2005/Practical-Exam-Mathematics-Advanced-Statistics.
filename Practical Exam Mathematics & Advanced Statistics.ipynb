```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
df = pd.read_excel('students_scores.xlsx', sheet_name='Student Scores')

# Step 1
math_mean = df['Math Score'].mean()
math_median = df['Math Score'].median()
math_mode = df['Math Score'].mode().tolist()

science_min = df['Science Score'].min()
science_max = df['Science Score'].max()
science_range = science_max - science_min
science_var = df['Science Score'].var(ddof=1) # sample variance
science_std = df['Science Score'].std(ddof=1)

print(f"Math Score - Mean: {math_mean:.2f}, Median: {math_median:.2f}, Mode: {math_mode}")
print(f"Science Score - Range: {science_range:.2f}, Variance: {science_var:.2f}, Std Dev: {science_std:.2f}")

# Step 2
pass_prob = (df['Pass Fail'] == 'Pass').mean()
print(f"Probability of passing: {pass_prob:.4f}")

df['Hours_Studied_gt_5'] = df['Hours Studied'] > 5
contingency_table = pd.crosstab(df['Pass Fail'], df['Hours_Studied_gt_5'], margins=True)
print("\nContingency Table:")
print(contingency_table)

# Conditional probability P(Pass | Hours_Studied > 5) = P(Pass and Hours > 5) / P(Hours > 5)
hours_gt_5 = df['Hours Studied'] > 5
passed_and_hours_gt_5 = ((df['Pass Fail'] == 'Pass') & hours_gt_5).sum()
total_hours_gt_5 = hours_gt_5.sum()
p_pass_given_hours_gt_5 = passed_and_hours_gt_5 / total_hours_gt_5 if total_hours_gt_5 > 0 else 0
print(f"P(Pass | Hours_Studied > 5): {p_pass_given_hours_gt_5:.4f}")

# Step 3
science_skew = df['Science Score'].skew()
science_kurt = df['Science Score'].kurtosis()
print(f"Science Score - Skewness: {science_skew:.4f}, Kurtosis: {science_kurt:.4f}")

# Step 4
v_math = df['Math Score'].head(5).values
v_science = df['Science Score'].head(5).values
print("\nFirst 5 Math Scores:", v_math)
print("First 5 Science Scores:", v_science)

dot_product = np.dot(v_math, v_science)
norm_1_math = np.linalg.norm(v_math, ord=1)
norm_2_math = np.linalg.norm(v_math, ord=2)

cos_angle = np.dot(v_math, v_science) / (np.linalg.norm(v_math) * np.linalg.norm(v_science))
angle_rad = np.arccos(np.clip(cos_angle, -1.0, 1.0))
angle_deg = np.degrees(angle_rad)

print(f"Dot Product: {dot_product:.2f}")
print(f"Norm 1 (Math): {norm_1_math:.2f}")
print(f"Norm 2 (Math): {norm_2_math:.2f}")
print(f"Angle between vectors (degrees): {angle_deg:.2f}")
```

```python
import numpy as np
import pandas as pd
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
file_path = 'students_scores.xlsx'
df = pd.read_excel(file_path, sheet_name='Student Scores')

print("="*60)
print("STEP 1: Measures of Central Tendency & Dispersion")
print("="*60)
math_mean = df['Math Score'].mean()
math_median = df['Math Score'].median()
math_mode = df['Math Score'].mode().tolist()
print(f"Math Score -> Mean: {math_mean:.2f} | Median: {math_median:.2f} | Mode: {math_mode}")

sci_range = df['Science Score'].max() - df['Science Score'].min()
sci_var = df['Science Score'].var(ddof=1)
sci_std = df['Science Score'].std(ddof=1)
print(f"Science Score -> Range: {sci_range:.2f} | Variance: {sci_var:.2f} | Std Dev: {sci_std:.2f}")

print("\n" + "="*60)
print("STEP 2: Probability Basics")
print("="*60)
pass_prob = (df['Pass Fail'] == 'Pass').mean()
print(f"Probability of Passing (P(Pass)): {pass_prob:.4f}")

df['Hours_Studied_gt_5'] = df['Hours Studied'] > 5
contingency_table = pd.crosstab(df['Pass Fail'], df['Hours_Studied_gt_5'], margins=True)
print("\nContingency Table (Pass/Fail vs Hours Studied > 5):")
print(contingency_table)

hours_gt_5 = df['Hours Studied'] > 5
passed_and_hours_gt_5 = ((df['Pass Fail'] == 'Pass') & hours_gt_5).sum()
total_hours_gt_5 = hours_gt_5.sum()
p_pass_given_hours_gt_5 = passed_and_hours_gt_5 / total_hours_gt_5 if total_hours_gt_5 > 0 else 0
print(f"\nConditional Probability P(Pass | Hours_Studied > 5): {p_pass_given_hours_gt_5:.4f}")

print("\n" + "="*60)
print("STEP 3: Distribution & Visualization")
print("="*60)
sci_skew = df['Science Score'].skew()
sci_kurt = df['Science Score'].kurtosis()
print(f"Science Score -> Skewness: {sci_skew:.4f} | Kurtosis: {sci_kurt:.4f}")

# 1. Histogram & Normal Curve for Math Score
plt.figure(figsize=(8, 5))
sns.histplot(df['Math Score'], kde=True, stat='density', color='skyblue', edgecolor='black')
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = stats.norm.pdf(x, math_mean, df['Math Score'].std())
plt.plot(x, p, 'r--', linewidth=2, label='Normal Fit')
plt.title('Histogram & Normal Curve for Math Score')
plt.xlabel('Math Score')
plt.ylabel('Density')
plt.legend()
plt.tight_layout()
plt.savefig('math_score_dist.png')
plt.close()

# 2. Q-Q Plot for English Score
plt.figure(figsize=(6, 6))
stats.probplot(df['English Score'], dist="norm", plot=plt)
plt.title('Q-Q Plot for English Score')
plt.tight_layout()
plt.savefig('english_score_qq.png')
plt.close()
print("Plots saved successfully as 'math_score_dist.png' and 'english_score_qq.png'.")

print("\n" + "="*60)
print("STEP 4: Linear Algebra Mini Task")
print("="*60)
v_math = df['Math Score'].head(5).values
v_science = df['Science Score'].head(5).values
print(f"Math Score Vector (First 5): {v_math}")
print(f"Science Score Vector (First 5): {v_science}")

dot_product = np.dot(v_math, v_science)
norm_1_math = np.linalg.norm(v_math, ord=1)
norm_2_math = np.linalg.norm(v_math, ord=2)

cos_angle = np.dot(v_math, v_science) / (np.linalg.norm(v_math) * np.linalg.norm(v_science))
angle_rad = np.arccos(np.clip(cos_angle, -1.0, 1.0))
angle_deg = np.degrees(angle_rad)

print(f"Dot Product: {dot_product:.2f}")
print(f"Norm 1 (Math Vector): {norm_1_math:.2f}")
print(f"Norm 2 (Math Vector): {norm_2_math:.2f}")
print(f"Angle between vectors: {angle_deg:.2f}°")
```
